```python
# List  packing
a = 25
b = 10.8
c = 'Hyd'
d = True
e = [a , b , c , d]
print(e)    # [25, 10.8, 'Hyd', True]
print(type(e)) # <class 'list'>
```

```python
# Find  outputs
list = [25 , 10.8 , 'Hyd' ,  True]
a , *b , c = list
print('a : ' , a)  #  a :  25
print('b : ' , b) #  b : [10.8 , 'Hyd']
print('c : ' , c) #  c :  True
print(type(b))  # <class 'list'>
x , *y = list
print('x : ' , x) # x :  25
print('y : ' , y)   # y :  [10.8, 'Hyd', True]
*p , q = list
print('p : ' , p) # p :  [25, 10.8, 'Hyd']
print('q : ' , q) # q :  True
```

```python
# Comparing  Lists
a = [10 , 20 , 15 , 18]
b = [10 , 20 , 15 , 18]
c = [10 , 20 , 25 , 9]
d = [10 , 20 , 7 , 22]
print(a == b)   # True
print(a  is   b)
print(a < c)  # True
print(a > d)  # True
print(a >= c)   # False
print(a <= d) # False
print(a != c)  # True
print(a != b) # False
print(a == c) # False
```
