```python
# Find  outputs  (Home  work)
def  square(fun):   # fun is num
	def  inner1():
		x = fun()   # num() # x=10
		return  x * x   # return 100
	return  inner1
def   double(fun):  # fun is inner1
	def  inner2():
		y = fun()   # inner1()      y=100
		return  2 * y   # retrun 200  (100*2=200)
	return   inner2
@double # num=double(square(num))   double(inner1)      num=inner2- i.e num points to inner2() func
@square
def  num():
	return  10
#end of the function
print(num())    # print(200)
''' output: 200'''
```

```python
# Find  outputs  (Home  work)
def   bold(fun):    # fun is inner2
	def  inner1():
		return  '<b>'  +  fun()  +  '</b>'  # <b>+inner2 result+</b>    <b><i><u>Hello world</u></i></b>
	return  inner1
def   italic(fun):  # fun is inner3
	def   inner2():
		return  '<i>'  +  fun() +  '</i>'   # <i>+inner3 result+</i>         <i><u>Hello world</u></i>
	return  inner2
def   underline(fun):       # fun is f1
	def   inner3():
		return  '<u>'  +  fun()  +  '</u>'  # <u>+result of f1+</u>             <u>Hello world</u>
	return  inner3
@bold       # f1=bold(italic(underline(f1)))        bold(italic(inner3))        bold(inner2)    f1=inner1 -i.e. ref f1 points to inner1
@italic
@underline
def   f1():
       return  'Hello  World'
# End  of  the  function
print(f1()) # print(<b><i><u>Hello world</u></i></b>)

'''output:
<b><i><u>Hello world</u></i></b>'''
```
