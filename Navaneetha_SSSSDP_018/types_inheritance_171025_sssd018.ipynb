```python
#  Multilevel  inheritance  demo  program
class  A:
	def    m1(self):
		print('class   A  method')
class  B(A):
	def  m1(self):
		print('class  B   method')
class   C(B):
	def  m1(self):
		print('class   C    method')
class   D(C):
	def   m1(self):
		print('class   D   method')
		C.m1(self) #How  to  call  method  m1()  of  class  C
		super().m1() #How  to  call  method  m1()  of  class  C  in  another  way  without  creating  an  object
		B.m1(self)  #How  to  call  method  m1()  of  class  B
		A.m1(self) #How  to  call  method  m1()  of  class  A
		super(A , self) . m1()
		super(C) . m1()
# End  of  the  class
d=D()
d.m1() #How  to  call  method  m1()  of  class  D
```

```python
# Parent  and  Child  constructor  demo  program  (Home  work)
class  parent:
	def   __init__(self , a1 , b1):
		self . a = a1
		self . b = b1
	def disp(self):
		print(self . a , self . b , sep = '\t' , end = '\t')
class   child(parent):
	def __init__(self , a2 = 0  , b2 = 0 , c2 = 0  , d2 = 0):
		super().__init__(a2,b2) #How  to  call  parent  class  constructor  with  a2 , b2
		self . c = c2
		self . d = d2
	def  disp(self):
		super().disp() #How  to  call  parent  class  disp()  method
		print(self . c , self . d , sep = '\t')
#end of the class
x = child(10 , 20 , 30 , 40)
y = child()
print('Object  x')
x . disp()
print('Object  y')
y . disp()

'''
Object  'x'  : a=10	 b=20	c=30  d=40

Object  'y'  : a=0  b=0  c=0  d=0

output:

Object x
10	20	30	40
Object y
0	0	0	0
'''
```

```python
# Parent class
class parent:
    a = 10  

    def __init__(self):
        self.x = 30  
        print('Parent constructor')

    def m1(self):
        print('Parent class instance method:', self.x)

    @classmethod
    def m2(cls):
        print('Parent class "class" method:', cls.a)
        print('Parent class "class" method (another way):', parent.a)

    @staticmethod
    def m3():
        print('Parent class static method:', parent.a)

    def __del__(self):
        print('Parent destructor:', getattr(self, 'x', None))
class child(parent):
    b = 20 
    def __init__(self):
        super().__init__() 
        self.y = 40         
        print('Child constructor')
    def m1(self):
        super().m1()      
        print('Child class instance method:', self.y)
    @classmethod
    def m2(cls):
        super(child, cls).m2()   
        cls.m2()                 
        print('Child class "class" method')
        print('Static variable a:', cls.a)
        print('Static variable a (another way):', parent.a)
        print('Static variable b:', cls.b)
        print('Static variable b (another way):', child.b)
    @staticmethod
    def m3():
        parent.m3()              
        print('Child class static method, static variable a:', parent.a)
        print('Child class static method, static variable b:', child.b)

    def __del__(self):
        super().__del__()       
        print('Child destructor:', getattr(self, 'y', None))


# --------------------------
# Calling methods

print('--- Creating object c ---')
c = child()           # calls constructors
print('--- Calling instance method m1 ---')
c.m1()                # calls child instance method (and parent via super)
print('--- Calling class method m2 ---')
child.m2()            # calls child class method
print('--- Calling static method m3 ---')
child.m3()            # calls child static method
print('--- Deleting object c ---')
del c                
```
