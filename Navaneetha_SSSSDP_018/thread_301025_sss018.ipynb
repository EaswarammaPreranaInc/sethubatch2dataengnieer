```python
'''
Modify  following   program  such  that  t1  should  execute  double()  function  and
t2  should  execute  square()  function
'''
from threading import *
import time
def   double():
	for   i  in  range(1 , 7):
		print('Double : ' , 2 * i)
		time . sleep(1)
def   square():
	for   i  in   range(1 , 7):
		print('Square : ' , i * i)
		time . sleep(1)
#  End  of  the  function
t1=Thread(target=double)
t2=Thread(target=square)
start = time . time()
t1.start()
t2.start()
t1.join()
t2.join()
end = time . time()
print(end - start) #  What  is   the   execution  time ?

'''Double :  2
Square :  1
Double :  4
Square :  4
Double : 6
Square :  9
Double : 8
Square :  16 
Square :  25
Double :  10
Double : 12
Square :  36
6.012'''
```
