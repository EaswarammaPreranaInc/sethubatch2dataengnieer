```python
import math

class Rat:
    # Do not modify the following methods (as requested)

    def get(self):  # Do not modify the method
        self.nr = int(input('Enter numerator : '))
        self.dr = int(input('Enter denominator : '))
        self.test()

    def test(self): # Do not modify the method
        while self.dr == 0:
            self.dr = int(input('Denominator can not be zero and re-enter :  '))

    def __str__(self):  # Do not modify the method
        return f'{self.nr} / {self.dr}'

    def simplify(self):  # Do not modify the method
        if self.nr != 0:
            # Note: math.gcd handles both positive and negative numbers.
            g = math.gcd(self.nr, self.dr)
            self.nr = self.nr // g
            self.dr = self.dr // g
        # Ensure denominator is positive for standard representation
        if self.dr < 0:
            self.nr = -self.nr
            self.dr = -self.dr

    # --- Start of Modified Operator Overloading Methods ---

    # Operator methods now return a *new* Rat object for the result, 
    # leaving the original operands (self and b) unchanged.

    def __add__(self, b):
        """Overloads the '+' operator for Rat objects (a + b)."""
        # (a/b) + (c/d) = (ad + bc) / bd
        new_nr = self.nr * b.dr + self.dr * b.nr
        new_dr = self.dr * b.dr

        result = Rat()
        result.nr = new_nr
        result.dr = new_dr
        result.simplify()
        return result

    def __sub__(self, b):
        """Overloads the '-' operator for Rat objects (a - b)."""
        # (a/b) - (c/d) = (ad - bc) / bd
        new_nr = self.nr * b.dr - self.dr * b.nr
        new_dr = self.dr * b.dr

        result = Rat()
        result.nr = new_nr
        result.dr = new_dr
        result.simplify()
        return result

    def __mul__(self, b):
        """Overloads the '*' operator for Rat objects (a * b)."""
        # (a/b) * (c/d) = ac / bd
        new_nr = self.nr * b.nr
        new_dr = self.dr * b.dr

        result = Rat()
        result.nr = new_nr
        result.dr = new_dr
        result.simplify()
        return result

    def __truediv__(self, b):
        """Overloads the '/' operator for Rat objects (a / b)."""
        # (a/b) / (c/d) = a * d / b * c
        new_nr = self.nr * b.dr
        new_dr = self.dr * b.nr

        result = Rat()
        result.nr = new_nr
        result.dr = new_dr

        # The calling code checks if b.nr != 0, preventing division by zero error here.
        if new_dr != 0:
            result.simplify()
        
        return result

# End of the class

# Modify the following statements to use the new operator overloads.
# This main logic remains the same as requested for testing.
a = Rat()
b = Rat()
a.get()
b.get()

# These operations now return new Rat objects (c, d, e, f) without changing a or b
c = a + b
d = a - b
e = a * b

print('\nResults:')
print('Sum :  ' , c)
print('Difference :  ' , d)
print('Product :  ' , e)

if b.nr != 0:
    f = a / b
    print('Division : ' , f)
else:
    print('Division is not permitted because the second numerator is 0.')
```

```python
'''
Overload   > ,  < ,  == ,  >=  , <=  , !=  on   Rational   class  objects

1) Let  object  'a'   contain   2 / 3  and   object  'b'  contain  5 / 9
    What  is  the  result  of  a > b ?  --->  True  due  to 18 > 15
    What  is  the  result  of  a < b ?  --->False  due  to  18  is  not  <  15
    What  is  the  result  of  a == b ?  --->	False  due  to  18  is  not  =  15
    What  is  the  result  of  a >= b ?  --->	True  due  to 18 >= 15
    What  is  the  result  of  a <= b ?  ---> 	False  due  to  18  is  not  <=  15
    What  is  the  result  of  a != b ?  ---> True  due  to 18 != 15

2) Imp  point  is  cross  product

3) What  is  the  method  call  to  _gt_()  method ?  --->  a > b
     What  is  the  method  call  to  _lt_()  method ?  ---> a < b
     What  is  the  method  call  to  _eq_()  method ?  --->  a == b
     What  is  the  method  call  to  _ge_()  method ?  --->  a >= b
     What  is  the  method  call  to  _le_()  method ?  --->  a <= b
     What  is  the  method  call  to  _ne_()  method ?  ---> a != b
'''
import math

class Rat:
	def get(self):
		self.nr = int(input('Enter numerator : '))
		self.dr = int(input('Enter denominator : '))

	def __gt__(self, b):
		# a > b if a.nr * b.dr > b.nr * a.dr
		return self.nr * b.dr > b.nr * self.dr

	def __lt__(self, b):
		return self.nr * b.dr < b.nr * self.dr

	def __eq__(self, b):
		return self.nr * b.dr == b.nr * self.dr

	def __ge__(self, b):
		return self.nr * b.dr >= b.nr * self.dr

	def __le__(self, b):
		return self.nr * b.dr <= b.nr * self.dr

	def __ne__(self, b):
		return self.nr * b.dr != b.nr * self.dr

#  End  of   the  class
a = Rat()
b = Rat()  # How to create two Rat class objects 'a' and 'b'
a.get()  # How to read 1st rational number into object 'a'
b.get()  # How to read 2nd rational number into object 'b'
if a > b:  # 1st rational is > 2nd rational number
	print('>')
if a < b:  # 1st rational is < 2nd rational number
	print('<')
if a == b:  # rational numbers are same
	print('==')
if a >= b:  # 1st rational is >= 2nd rational number
	print('>=')
if a <= b:  # 1st rational is <= 2nd rational number
	print('<=')
if a != b:  # rational numbers are different
	print('!=')
```

Create a class Box with private data members length, breadth, and height. Add a method to calculate the volume.

```python
class Box():
    def __init__(self,length,breadth,height):
        self.length=length
        self.breadth=breadth
        self.height=height
    def volume(self):
        return self.length*self.breadth*self.height
#end of the class
c = Box(10,20,30)
print(c.volume())
```

Write a program to create a copy constructor that copies the values from one object to another.

```python
class Student:
    def __init__(self, name=None, roll_no=None):
        self.name = name
        self.roll_no = roll_no

    # copy constructor
    @classmethod
    def copy(cls, other):
        return cls(other.name, other.roll_no)

# original object
s1 = Student("John", 101)

# copied object
s2 = Student.copy(s1)

# display both
print("Original:", s1.name, s1.roll_no)
print("Copy:", s2.name, s2.roll_no)
```

Create a class Employee that has a destructor displaying a message when the object is destroyed.

```python
class Employee:
    def __init__(self, name, emp_id,salary):
        self.name = name
        self.emp_id = emp_id
        self.salary = salary
    def disp(self):
        print(f"Name: {self.name}, ID: {self.emp_id}, Salary: {self.salary}")
    def __del__(self):
        print(f"Employee {self.name} with ID {self.emp_id} is being deleted.")

# create object
a = Employee("John", 101, 50000)

# display details
a.disp()
del a

```
