```python
'''Write a function for gcd(a,b) without using any predefined function Use the function to 
print gcd of 3 numbers 
Inputs: a,b and c 
eg: gcd(12,15)=3'''

def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a
x = int(input("Enter first number: "))
y = int(input("Enter second number: "))
z = int(input("Enter third number: "))
result = gcd(gcd(x, y), z)

print("GCD of", x, ",", y, "and", z, "is:", result)
```

```python
'''Write a function to determine index of each largest element in the list.(Assume that largest element is element is in multiple location of the list)
Function should return a list of all indexes of largest element '''
def largest_indexes(lst):
    largest = lst[0]
    for num in lst:
        if num > largest:
            largest = num
    indexes = []
    for i in range(len(lst)):
        if lst[i] == largest:
            indexes.append(i)
    
    return indexes
n = int(input("Enter number of elements: "))
lst = []

for i in range(n):
    val = int(input(f"Enter element {i+1}: "))
    lst.append(val)

print("List:", lst)
print("Indexes of largest element:", largest_indexes(lst))
```

```python
'''Write a function to count number of characters,vowels,consonants,spaces,tabs, words in a string
Function should return a list of 6 results '''
def count_all(s):
    vowels_set = "aeiouAEIOU"
    
    characters = len(s)         
    vowels = 0
    consonants = 0
    spaces = 0
    tabs = 0
    
    for ch in s:
        if ch in vowels_set:
            vowels += 1
        elif ch.isalpha():       
            consonants += 1
        elif ch == " ":
            spaces += 1
        elif ch == "\t":
            tabs += 1
    
    words = len(s.split())      
    
    return [characters, vowels, consonants, spaces, tabs, words]
text = input("Enter a string: ")
result = count_all(text)

print("Characters:", result[0])
print("Vowels:", result[1])
print("Consonants:", result[2])
print("Spaces:", result[3])
print("Tabs:", result[4])
print("Words:", result[5])
```

```python
'''Write a program to determine largest and smallest element of inputs passed from coomand line
Eg: py filename.py 10 20 5 15 30 3 25 40 
largest element: 40
smallest element: 3
donot use any predefined function or method to determine largest and smallest '''
import sys
numbers = [int(x) for x in sys.argv[1:]]
largest = numbers[0]
smallest = numbers[0]
for num in numbers[1:]:
    if num > largest:
        largest = num
    if num < smallest:
        smallest = num
print("Largest element:", largest)
print("Smallest element:", smallest)
```

```python
'''Write a function to test a matrix is symmeteric or not 
Function should return True/False
input: 10 20 30
       20 40 50
       30 50 60
Output: Symmetric matrix'''
def is_symmetric(matrix):
    rows = len(matrix)
    cols = len(matrix[0])
    if rows != cols:
        return False
    for i in range(rows):
        for j in range(cols):
            if matrix[i][j] != matrix[j][i]:
                return False
    return True
n = int(input("Enter the size of square matrix (n): "))
matrix = []

print("Enter the elements row by row:")
for i in range(n):
    row = list(map(int, input().split()))
    matrix.append(row)
print( is_symmetric(matrix))
```

```python
'''Write a program to print pascals triangle'''
rows = int(input("Enter number of rows: "))

for i in range(rows):
    print(" " * (rows - i), end="")
    
    num = 1
    for j in range(i + 1):
        print(num, end=" ")
        num = num * (i - j) // (j + 1)
    print()
```

```python
'''Write a function to convert binary number to decimal number without any predefined function or method'''
def decimal(str):
    decimal = 0
    power = 0
    for bit in str[::-1]:
        if bit == '1':
            decimal += 2 ** power
        power += 1
    
    return decimal
binary = input("Enter a binary number: ")
print("Decimal value:", decimal(binary))
```
