```python
#  Write  a  program  to  convert  postfix  to  prefix
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        return self.items.pop()

    def peek(self):
        return self.items[-1]

    def is_empty(self):
        return len(self.items) == 0


def postfix_to_prefix(postfix):
    s = Stack()

    for ch in postfix:
        if ch.isalnum():   
            s.push(ch)
        else:             
            op2 = s.pop()
            op1 = s.pop()
            s.push(ch + op1 + op2)

    return s.pop()

postfix = input("Enter postfix expression: ")   
prefix = postfix_to_prefix(postfix)
print("Prefix expression:", prefix)
```

```python
#  Write  a  program  to  convert  prefix  to  postfix

class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        return self.items.pop()

    def peek(self):
        return self.items[-1]

    def is_empty(self):
        return len(self.items) == 0


def prefix_to_postfix(prefix):
    prefix=prefix[::-1]
    s = Stack()

    for ch in prefix:
        if ch.isalnum():   
            s.push(ch)
        else:             
            op1 = s.pop()
            op2 = s.pop()
            s.push(op1 + op2+ch)

    return s.pop()

prefix = input("Enter prefix expression: ")   
postfix = prefix_to_postfix(prefix)
print("Postfix expression:", postfix)
```
