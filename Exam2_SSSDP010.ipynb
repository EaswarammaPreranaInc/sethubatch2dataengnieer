```python
'''Write a function for power(a,b) without using any predefined function or ** operator 
Use the function to print a^b^c 
Inputs are a,b,c 
Hint: power can be +ve,-ve or 0'''
def power(a, b):
    res = 1
    if b > 0:                      
        for i in range(b):
            res *= a
    elif b < 0:                    
        for i in range(-b):
            res *= a
        res = 1 / res
    else:                         
        res = 1
    return res
a = float(input("Enter base a: "))
b = int(input("Enter exponent b: "))
c = int(input("Enter exponent c: "))
d=power(b,c)
result = power(a,d)
print(f"{a}^{b}^{c} = {result}")
```

```python
'''Write a function to test two strings are anagram or not 
Function should return True/False '''
def anagram(str1, str2):
    str1 = str1.replace(" ", "").lower()
    str2 = str2.replace(" ", "").lower()
    return sorted(str1) == sorted(str2)
s1 = input("Enter first string: ")
s2 = input("Enter second string: ")

result = anagram(s1, s2)

print( result)
   
```

```python
'''Write a function to return a list of 3 elements i.e Largest element,row number and column number of a matrix
Use function to print the results  '''
def find_largest(matrix):
    largest = matrix[0][0]
    row_index = 0
    col_index = 0
    for i in range(len(matrix)):         
        for j in range(len(matrix[i])):   
            if matrix[i][j] > largest:
                largest = matrix[i][j]
                row_index = i
                col_index = j

    return [largest, row_index, col_index]
rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of columns: "))
matrix = []
print("Enter matrix elements row by row:")
for i in range(rows):
    row = list(map(int, input().split()))
    matrix.append(row)
result = find_largest(matrix)
print("Largest element, Row index, Column index:", result)
```

```python
'''Write a python program to print the pattern. No of rows will be the input paramter
ABCDEFGFEDCBA
ABCDEF FEDBCA
ABCDE   EDBCA
ABCD     DCBA
ABC       CBA
AB         BA
A           A'''
def pattern(n):
    for i in range(n):
        left = ''.join(chr(65 + j) for j in range(n - i))
        right = ''.join(chr(65 + j) for j in range(n - i - 1, -1, -1))
        spaces = ' ' * (2 * i - 1) if i > 0 else ''
        if i == 0:
            print(left + right[1:])  
        else:
            print(left + spaces + right)
rows = int(input("Enter number of rows: "))
pattern(rows)
```

```python
'''Write a function to convert decimal number to binary number without any pre-defined function or method'''
def bin(a):
    if a==0:
        return 0
    binary=""
    while a>0:
        rem=a%2
        binary=str(rem)+binary
        a=a//2
    return binary
num=int(input("Enter any number:"))
print("Binary Number:",bin(num))
```
