```python
# Modify  following  div  function  such  that  div(9 , 2)   and   div(2 , 9)  should  return  4.5  only
def decor(fun):
    def inner(a, b):
        big = max(a, b)
        small = min(a, b)
        return fun(big, small)
    return inner#How  to  decorate  the  function  such  that  4.5  is  returned
@decor
def div(a, b):
    return a / b

print(div(9, 2))  
print(div(2, 9))  
```

```python
#  Most  tricky   program
# Same  decorator  to  multiple  functions  with  different  signatures
def   decor(fun):
	print(fun . __name__)#f1 f2 f3 f4
	def   inner(*x):  #   *  is  var-arg  parameter
		print(x)#(10,)
		fun(*x)  #  *  unpacks  object  'x'
		print('End  of  decoration')
	return  inner
@decor#f1=decor(f1)
def   f1(x):
	print('f1  function  :  ' , x)#f1  function  :   10
@decor#f2=decor(f2)
def   f2(x , y):
	print('f2  function  :  ' , x , y )
@decor#f3=decor(f3)
def  f3(x , y , z):
	print('f3 function : ' , x , y , z)
@decor##f4=decor(f4)
def   f4():
	print('f4 function')
# end of function
f1(10)
f2(25 , 10.8)
f3('Hyd' ,  True  , 3 + 4j)
f4()
'''Outputs:f1
f2
f3
f4
(10,)
f1  function  :   10
End  of  decoration
(25, 10.8)
f2  function  :   25 10.8
End  of  decoration
('Hyd', True, (3+4j))
f3 function :  Hyd True (3+4j)
End  of  decoration
()
f4 function
End  of  decoration'''
```

```python
# Find  outputs  (Home  work)
def  square(fun):   #fun is num
	def  inner1():  
		x = fun()   # x=num()   # x=10
		return  x * x   # return 100
	return  inner1
def   double(fun):  # fun is inner1
	def  inner2():
		y = fun()   # y=inner1()    # y=100
		return  2 * y   # 2*100=200
	return   inner2
@double     # num=double(square(num))   # double(inner1)        # num=inner2 i.e ref num is point to inner2
@square
def  num():
	return  10
#end of the function
print(num())    # print(200)
'''200'''
```

```python
# Find  outputs  (Home  work)
def   bold(fun):    # fun is inner2
	def  inner1():
		return  '<b>'  +  fun()  +  '</b>'  # <b><i><u>Hello World</u></i></b>
	return  inner1
def   italic(fun):  # fun is inner3
	def   inner2():
		return  '<i>'  +  fun() +  '</i>'   # <i><u>Hello World</u></i>
	return  inner2
def   underline(fun):   # fun is f1
	def   inner3():
		return  '<u>'  +  fun()  +  '</u>'  # <u>Hello World</u>
	return  inner3
@bold   # f1= bold(italic(underline(f1)))   bold(italic(inner3))    bold(inner2)    f1=inner1 i.e ref f1 points to inner1
@italic
@underline
def   f1():
       return  'Hello  World'
# End  of  the  function
print(f1()) # print(<b><i><u>Hello World</u></i></b>)

'''<b><i><u>Hello World</u></i></b>'''
```
