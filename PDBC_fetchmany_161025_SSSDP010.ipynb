```python
'''
insertion multiple rows into student table
'''
import mysql.connector as mc
try:
    con = mc.connect(
        database='student',
        user='root',
        password='Prasanna@1982',
        host='127.0.0.1'
    )
    cur = con.cursor()
    n = int(input('How many rows do you want to insert ? : '))
    student_list = []  # list of tuples
    for i in range(n):
        print(f'Student {i+1}')
        student_id = int(input('Enter student number : '))
        first_name= input('Enter First name : ')
        last_name = input('Enter Last name : ')
        date_of_birth = input('Enter Date of Birth(YYYY-MM-DD) : ')
        gender = input('Enter gender (M/F): ')
        email=input('Enter email')
        student_list.append((student_id, first_name, last_name, date_of_birth, gender,email)) 
    cur.executemany(
        'INSERT INTO student (student_id, first_name, last_name, date_of_birth, gender,email) VALUES (%s, %s, %s, %s, %s, %s)',
        student_list
    )
    con.commit()
    print(f'{cur.rowcount} rows are inserted')
    input('Press any key to continue...')
    cur.close()
    con.close()
except mc.errors.IntegrityError:
    print('Duplicate empno detected empno must be unique (PRIMARY KEY).')
except mc.errors.DatabaseError:
    print('Please start MySQL')
except ValueError:
    print('Invalid input! Please enter correct data types.')
```
